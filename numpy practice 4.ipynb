```python
import numpy as np

zeros = np.zeros((3, 4))
ones = np.ones((3, 4))
eye = np.eye(4)

arrays = {
    "Zeros": zeros,
    "Ones": ones,
    "Eye": eye
}

for name, arr in arrays.items():
    print(name)
    print("Shape :", arr.shape)
    print("Size  :", arr.size)
    print("Ndim  :", arr.ndim)
    print("Dtype :", arr.dtype)
    print()

print("Eye matrix row by row:")
for row in eye:
    for value in row:
        print(int(value), end=" ")
    print()

eye_copy = eye.copy()

print("\nIndependent copy?")
print(eye is eye_copy)
```

```python
import numpy as np

zeros = np.zeros((3, 4))
ones = np.ones((3, 4))
eye = np.eye(4)

print("Zeros")
print(zeros)
print("Shape:", zeros.shape)
print("Size:", zeros.size)
print("Ndim:", zeros.ndim)
print("Dtype:", zeros.dtype)

print("\nOnes")
print(ones)
print("Shape:", ones.shape)
print("Size:", ones.size)
print("Ndim:", ones.ndim)
print("Dtype:", ones.dtype)

print("\nEye")
print(eye)
print("Shape:", eye.shape)
print("Size:", eye.size)
print("Ndim:", eye.ndim)
print("Dtype:", eye.dtype)

print("\nEye matrix row by row")
for i in range(4):
    for j in range(4):
        print(int(eye[i][j]), end=" ")
    print()

copy_eye = eye.copy()

print("\nIs it an independent copy?")
print(copy_eye is eye)
```

```python
import numpy as np

one_d = np.array([10, 20, 30, 40, 50])

two_d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

three_d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array")
print(one_d)
print("Shape:", one_d.shape)
print("Size:", one_d.size)
print("Ndim:", one_d.ndim)
print("Dtype:", one_d.dtype)

print("\n2D Array")
print(two_d)
print("Shape:", two_d.shape)
print("Size:", two_d.size)
print("Ndim:", two_d.ndim)
print("Dtype:", two_d.dtype)

print("\n3D Array")
print(three_d)
print("Shape:", three_d.shape)
print("Size:", three_d.size)
print("Ndim:", three_d.ndim)
print("Dtype:", three_d.dtype)
```
