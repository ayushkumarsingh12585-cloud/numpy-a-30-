```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

copy_arr = arr.copy()
same_arr = arr

# copy() creates an independent copy
# same_arr is another reference to the same array

print(copy_arr is arr)
print(same_arr is arr)
```
